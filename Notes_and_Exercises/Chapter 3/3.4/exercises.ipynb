## Q4

Compute Explosion: It is $O(N^3)$ to calc $\Delta \theta = -H^{-1} \nabla L$ because of inversion of $H$.

> Why the form is $\Delta \theta = -H^{-1} \nabla L$?
>
> Consider second-order Taylor formula:
> 
> $$ L(\theta + \Delta \theta) \approx L(\theta) + \nabla L(\theta)^T \Delta \theta + \frac{1}{2} \Delta\theta^T H \Delta \theta $$
>
> Our goal is to find a $\Delta \theta$, minimizing the $L$. Then we just need to calc the derivative of $L(\theta + \Delta \theta)$ and get that $\Delta \theta = -H^{-1} \nabla L$.


To reach $H^{-1}$ fast, we can use Quasi-Newton method.
