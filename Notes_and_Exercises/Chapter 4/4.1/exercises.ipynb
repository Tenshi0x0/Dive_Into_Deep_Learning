## Q1

As $\hat{y}_j = \frac{\exp(o_j)}{\sum_i \exp(o_i)}$, we have

$$
l(y, \hat{y}) = - \sum_j y_j o_j + \log \sum_i \exp(o_i)
$$

Thus, first derivative is:

$$
\frac{\partial l}{\partial o_j} = -y_j + \frac{\exp(o_j)}{\sum_i \exp(o_i)} = \hat{y_j} - y_j.
$$

Then, second derivative is,

> Let $S = \sum_k \exp(o_k)$.

i> $i=j$:
$$
H_{ii} = \frac{\exp(o_i)S - \exp(o_i)^2}{S^2}.
$$

ii> $i \ne j$:
$$
H_{ij}=-\frac{\exp(o_i + o_j)}{S^2}.
$$

Overall, $H = \text{diag}(\hat{y}) - \hat{y}\hat{y}^T.$

For cov, let $X=e_C$, here $C \sim \text{Categorical}(p_i)$:

$$
\text{Cov}(X) = E[XX^T] - E[X]E[X^T] = \text{diag}(\hat{y}) - \hat{y}\hat{y}^T.
$$

## Q5

### (1)

Let $d = \log(\exp(a) + \exp(b)) - \max(a, b)$, consider $\exp(d)$:

$$
\exp(d) = \frac{\exp(a) + \exp(b)}{\exp(\max(a, b))}
= 1 + \exp(\min(a, b) - \max(a, b)) > 1.
$$

Thus $d > 0$.

### (2)

Refer to (1), the difference is

$$
d=\log(1+\exp(\min(a,b)-\max(a,b))).
$$

To make it as small as possible, we need to make $\max(a,b)-\min(a,b)$ as large as possible. Without loss of generality, let $b=0$ and $a\ge b$. Then

$$
d=\log(1+\exp(-a)).
$$

As $a\to\infty$, we have $d\to 0$. Therefore, the difference can approach $0$, but it is always positive for finite $a,b$.

### (3)

Let

$$
d_\lambda = \lambda^{-1}\mathrm{RealSoftMax}(\lambda a, \lambda b) - \max(a, b).
$$

Then,

$$
\lambda d_\lambda
= \log(\exp(\lambda a)+\exp(\lambda b))-\lambda\max(a,b).
$$

Consider $\exp(\lambda d_\lambda)$:

$$
\begin{aligned}
\exp(\lambda d_\lambda)
&=
\frac{\exp(\lambda a)+\exp(\lambda b)}
{\exp(\lambda\max(a,b))} \\
&=
1+\exp(\lambda(\min(a,b)-\max(a,b)))>1.
\end{aligned}
$$

Thus,

$$
\lambda d_\lambda > 0.
$$

Since $\lambda>0$, we have

$$
d_\lambda>0.
$$

Therefore,

$$
\lambda^{-1}\mathrm{RealSoftMax}(\lambda a,\lambda b)>\max(a,b).
$$

### (4)

From (3), the difference is

$$
d_\lambda
=
\frac{1}{\lambda}
\log\left(1+\exp(\lambda(\min(a,b)-\max(a,b)))\right).
$$

Equivalently,

$$
d_\lambda
=
\frac{1}{\lambda}
\log\left(1+\exp(-\lambda |a-b|)\right).
$$

When $\lambda\to\infty$, we have

$$
\exp(-\lambda |a-b|)\to 0
$$

if $a\ne b$. Thus,

$$
d_\lambda\to 0.
$$

If $a=b$, then

$$
d_\lambda=\frac{\log 2}{\lambda}\to 0.
$$

Therefore,

$$
\lambda^{-1}\mathrm{RealSoftMax}(\lambda a,\lambda b)\to \max(a,b).
$$

### (5)

An analogous softmin function can be constructed as

$$
\mathrm{RealSoftMin}(a,b)
=
-\log(\exp(-a)+\exp(-b)).
$$

Let

$$
d=\min(a,b)-\mathrm{RealSoftMin}(a,b).
$$

Then,

$$
\begin{aligned}
\exp(d)
&=
\frac{\exp(\min(a,b))}
{\exp(\mathrm{RealSoftMin}(a,b))} \\
&=
\exp(\min(a,b))(\exp(-a)+\exp(-b)) \\
&=
1+\exp(\min(a,b)-\max(a,b))>1.
\end{aligned}
$$

So $d>0$, and hence

$$
\mathrm{RealSoftMin}(a,b)<\min(a,b).
$$

The scaled version is

$$
-\frac{1}{\lambda}\log(\exp(-\lambda a)+\exp(-\lambda b)).
$$

When $\lambda\to\infty$, it converges to

$$
\min(a,b).
$$

### (6)

For more than two numbers $a_1,\dots,a_n$, we can define

$$
\mathrm{RealSoftMax}(a_1,\dots,a_n)
=
\log\sum_{i=1}^n \exp(a_i).
$$

The scaled version is

$$
\frac{1}{\lambda}\log\sum_{i=1}^n \exp(\lambda a_i).
$$

Let

$$
M=\max_i a_i.
$$

Then,

$$
\begin{aligned}
\frac{1}{\lambda}\log\sum_{i=1}^n \exp(\lambda a_i)
&=
\frac{1}{\lambda}\log\left(\exp(\lambda M)\sum_{i=1}^n \exp(\lambda(a_i-M))\right) \\
&=
M+\frac{1}{\lambda}\log\sum_{i=1}^n \exp(\lambda(a_i-M)).
\end{aligned}
$$

Since $a_i-M\le 0$, the second term goes to $0$ as $\lambda\to\infty$. Thus,

$$
\frac{1}{\lambda}\log\sum_{i=1}^n \exp(\lambda a_i)
\to
\max_i a_i.
$$

Similarly,

$$
\mathrm{RealSoftMin}(a_1,\dots,a_n)
=
-\log\sum_{i=1}^n \exp(-a_i),
$$

and the scaled version

$$
-\frac{1}{\lambda}\log\sum_{i=1}^n \exp(-\lambda a_i)
\to
\min_i a_i.
$$
