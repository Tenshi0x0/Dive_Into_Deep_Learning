## Q1

Let $n$ be number of samples and $B$ be size of one batch (not the last one). Then the number of batches is $m = \lceil n/B \rceil$.

And denote the avg value of pre $m-1$ block as $v$, we have



$$
sum = B(m-1)v + [n-(m-1)B]l_v^b \\

L_v = sum/n \\

L_v^q = \frac{(m-1)v + l_v^b}{m}
$$

Eliminate the $v$, we get

$$
L_v = \frac{mBL_v^q + (n-mB)l_v^b}{n} = \frac{mB}{n}L_v^q + (1 - \frac{mB}{n})l_v^b. \tag{1}
$$

It can be considered as a linear equation about $\frac{mB}{n}$ and easy to find the solution is $mB=n$ iff $L_v = L_v^q$.

## Q2

For $E[L_v] = E[l_v^b]$, substitute this into equation (1) and we find $E[L_v] = E[L_v^q]$;

Still prefer $L_v$ for smaller variance:

With iid losses ($\mathrm{Var} = \sigma^2$), any weighted average $T = \sum_i w_i \ell_i$, $\sum_i w_i = 1$ has

$$
\mathrm{Var}(T) = \sigma^2 \sum_i w_i^2 \ge \sigma^2/n
$$

by Cauchy–Schwarz, equality iff $w_i = 1/n$, i.e. $L_v$. But $L_v^q$ weights the last batch's samples by $\frac{1}{mb'} > \frac{1}{n}$, so

$$
\mathrm{Var}(L_v^q) = \frac{\sigma^2}{m^2}\left(\frac{m-1}{B} + \frac{1}{b'}\right) > \frac{\sigma^2}{n} = \mathrm{Var}(L_v) \quad (b' < B).
$$
